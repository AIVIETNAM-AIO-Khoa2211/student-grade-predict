# 1. Set up

```python
import sys
from pathlib import Path

# Thêm thư mục gốc project (cha của notebooks/) vào sys.path
project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.config import DATA_RAW_DIR, DATA_PROCESSED_DIR, DATA_FILTERED_DIR, RESULTS_DIR
from src.config import DECISION_TREE_PARAMS, RANDOM_FOREST_PARAMS, ADABOOST_PARAMS
from src.config import DECISION_TREE_FE_PARAMS, RANDOM_FOREST_FE_PARAMS, ADABOOST_FE_PARAMS
from src.config import DECISION_TREE_FI_PARAMS, RANDOM_FOREST_FI_PARAMS, ADABOOST_FI_PARAMS
from src.config import LOG_REG_PARAM #Tham số này tối ưu với cả 3 loại dữ liệu

from src.data.loader import read_csv
from src.models import concat_and_train_DT,concat_and_train_RF,concat_and_train_AB, concat_and_train_logistic
from src.evaluation.metrics import evaluate_test
```

# 2.Load data

## 2.1 Load original data

```python
X_train, y_train = read_csv(DATA_RAW_DIR/"train.csv")
X_val, y_val = read_csv(DATA_RAW_DIR/"val.csv")
X_test, y_test = read_csv(DATA_RAW_DIR/"test.csv")
```

## 2.2 Load featured data

```python
X_train_fe, y_train_fe = read_csv(DATA_PROCESSED_DIR/"train.csv")
X_val_fe, y_val_fe = read_csv(DATA_PROCESSED_DIR/"val.csv")
X_test_fe, y_test_fe = read_csv(DATA_PROCESSED_DIR/"test.csv")
```

## 2.3 Load filtered data

```python
X_train_fi, y_train_fi = read_csv(DATA_FILTERED_DIR/"train.csv")
X_val_fi, y_val_fi = read_csv(DATA_FILTERED_DIR/"val.csv")
X_test_fi, y_test_fi = read_csv(DATA_FILTERED_DIR/"test.csv")
```

# 3. Train model

## 3.1 Train model on original data

```python
tree_raw = concat_and_train_DT(X_train, y_train, X_val, y_val, **DECISION_TREE_PARAMS)
rf_raw = concat_and_train_RF(X_train, y_train, X_val, y_val, **RANDOM_FOREST_PARAMS)
ada_raw = concat_and_train_AB(X_train, y_train, X_val, y_val, **ADABOOST_PARAMS)
log_raw, scaler_raw = concat_and_train_logistic(X_train, y_train, X_val, y_val, **LOG_REG_PARAM)
```

## 3.2 Train model on fe data

```python
tree_fe = concat_and_train_DT(X_train_fe, y_train_fe, X_val_fe, y_val_fe, **DECISION_TREE_FE_PARAMS)
rf_fe = concat_and_train_RF(X_train_fe, y_train_fe, X_val_fe, y_val_fe, **RANDOM_FOREST_FE_PARAMS)
ada_fe = concat_and_train_AB(X_train_fe, y_train_fe, X_val_fe, y_val_fe, **ADABOOST_FE_PARAMS)
log_fe, scaler_fe = concat_and_train_logistic(X_train_fe, y_train_fe, X_val_fe, y_val_fe, **LOG_REG_PARAM)
```

## 3.3 Train model on FI data

```python
tree_fi = concat_and_train_DT(X_train_fi, y_train_fi, X_val_fi, y_val_fi, **DECISION_TREE_FI_PARAMS)
rf_fi = concat_and_train_RF(X_train_fi, y_train_fi, X_val_fi, y_val_fi, **RANDOM_FOREST_FI_PARAMS)
ada_fi = concat_and_train_AB(X_train_fi, y_train_fi, X_val_fi, y_val_fi, **ADABOOST_FI_PARAMS)
log_fi, scaler_fi = concat_and_train_logistic(X_train_fi, y_train_fi, X_val_fi, y_val_fi, **LOG_REG_PARAM)
```

# 4. Evaluate on test

```python
# --- Đánh giá 4 model RAW trên X_test, y_test ---
tree_raw_test_acc = evaluate_test(tree_raw, X_test, y_test, model_name="Decision Tree (raw)")
rf_raw_test_acc = evaluate_test(rf_raw, X_test, y_test, model_name="Random Forest (raw)")
ada_raw_test_acc = evaluate_test(ada_raw, X_test, y_test, model_name="AdaBoost (raw)")
log_raw_test_acc = evaluate_test(log_raw, scaler_raw.transform(X_test), y_test, model_name="Logistic Regression (raw)")

# --- Đánh giá 4 model FEATURE ENGINEERED trên X_test_fe, y_test_fe ---
tree_fe_test_acc = evaluate_test(tree_fe, X_test_fe, y_test_fe, model_name="Decision Tree (FE)")
rf_fe_test_acc = evaluate_test(rf_fe, X_test_fe, y_test_fe, model_name="Random Forest (FE)")
ada_fe_test_acc = evaluate_test(ada_fe, X_test_fe, y_test_fe, model_name="AdaBoost (FE)")
log_fe_test_acc = evaluate_test(log_fe, scaler_fe.transform(X_test_fe), y_test_fe, model_name="Logistic Regression (FE)")

# --- Đánh giá 4 model FEATURE IMPORTANCE (Boruta-filtered) trên X_test_fi, y_test_fi ---
tree_fi_test_acc = evaluate_test(tree_fi, X_test_fi, y_test_fi, model_name="Decision Tree (FI)")
rf_fi_test_acc = evaluate_test(rf_fi, X_test_fi, y_test_fi, model_name="Random Forest (FI)")
ada_fi_test_acc = evaluate_test(ada_fi, X_test_fi, y_test_fi, model_name="AdaBoost (FI)")
log_fi_test_acc = evaluate_test(log_fi, scaler_fi.transform(X_test_fi), y_test_fi, model_name="Logistic Regression (FI)")
```

```python
import pandas as pd

results = pd.DataFrame({
    "model": ["Decision Tree", "Random Forest", "AdaBoost", "Logistic Regression"] * 3,
    "feature_set": ["raw"] * 4 + ["feature_engineered"] * 4 + ["feature_filter"] * 4,
    "test_accuracy": [
        tree_raw_test_acc, rf_raw_test_acc, ada_raw_test_acc, log_raw_test_acc,
        tree_fe_test_acc, rf_fe_test_acc, ada_fe_test_acc, log_fe_test_acc,
        tree_fi_test_acc, rf_fi_test_acc, ada_fi_test_acc, log_fi_test_acc,
    ],
}).sort_values("test_accuracy", ascending=False).reset_index(drop=True)

print(results)
```
